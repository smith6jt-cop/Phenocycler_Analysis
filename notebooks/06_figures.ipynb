# 6 — Cell type × marker identity figures

The final identity QC: does each `cell_type` express its defining markers? Exact per-`cell_type` means over
all typed cells (one streaming pass per donor, no subsampling):

- **Dotplot** — `cell_type` × the RESTORE gate markers. Dot **size** = % of cells `_pos`; dot **colour** =
  mean RESTORE `_norm` on a single shared **absolute** scale. Each cell_type lights up its own markers
  (Endocrine-β→INS, Endocrine-α→GCG, T→CD3e, B→CD20, Macrophage→CD68/CD163, Endothelial→CD31, Muscle→SMA,
  Fibroblast→Vimentin).
- **Heatmap** — `cell_type` × the full REDSEA-corrected TYPE-marker panel, z-scored per marker.

```python
# Parameters for this step (self-contained -- no config.ini). Edit the paths for your machine.
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))   # make `phenocycler` importable from notebooks/
from phenocycler import PipelineConfig

REPO = pathlib.Path.cwd().resolve().parents[1]        # Islet-Explorer-Senior (parent of the submodule; data/ lives here)
# Run on a subset of donors (None = every donor under data/cells/donor_id=*).
DONORS = None            # e.g. ["6374", "6380"] to iterate on a few


cfg = PipelineConfig(
    data_dir       = REPO / "data",
    donor_metadata = pathlib.Path("/home/smith6jt/IO60panc2nd/donor_metadata_panc.xlsx"),
)
print("data_dir     :", cfg.data_dir)
print("phenotype_dir:", cfg.phenotype_dir)
donors = DONORS or cfg.discover_donors()
print(f"donors: {len(donors)} " + ("(subset)" if DONORS else "(all)") + f" -> {donors[:6]}" + (" ..." if len(donors) > 6 else ""))
```

```python
from phenocycler.figures import run_figures
run_figures(cfg, donors=DONORS)   # donors=None -> all
```

```python
from IPython.display import Image, display
for fig in (cfg.phenotype_dir / "celltype_marker_dotplot.png",
            cfg.phenotype_dir / "celltype_marker_heatmap.png"):
    if fig.exists():
        display(Image(filename=str(fig)))
    else:
        print(f"({fig.name} not found — run the figures cell above)")
```
